```python
import pandas as pd
import numpy as np
import os

cmapss_path = "/kaggle/input/nasa-turbofan-engine-degradation-simulation/"

cols = ['engine_id', 'cycle'] + \
       [f'op_setting_{i}' for i in range(1, 4)] + \
       [f'sensor_{i}' for i in range(1, 22)]

train_df = pd.read_csv(cmapss_path + "train_FD001.txt", sep=r"\s+", header=None, names=cols)
test_df  = pd.read_csv(cmapss_path + "test_FD001.txt",  sep=r"\s+", header=None, names=cols)
rul_df   = pd.read_csv(cmapss_path + "RUL_FD001.txt",   header=None, names=["RUL"])

print(train_df.shape, test_df.shape, rul_df.shape)
```

```python
ai4i_path = "/kaggle/input/predictive-maintenance-dataset/ai 2020.csv"
ai4i_df = pd.read_csv(ai4i_path)

print(ai4i_df.shape)
ai4i_df.head()
```

```python
skab_base_path = "/kaggle/input/skoltech-anomaly-benchmark-skab/"

skab_dfs = []

for root, dirs, files in os.walk(skab_base_path):
    for file in files:
        if file.endswith(".csv"):
            full_path = os.path.join(root, file)
            df = pd.read_csv(full_path)
            df["source_file"] = file
            skab_dfs.append(df)

print("Total SKAB files loaded:", len(skab_dfs))

skab_df = pd.concat(skab_dfs, ignore_index=True)
print("SKAB shape:", skab_df.shape)
```

```python
print("C-MAPSS missing values:", train_df.isna().sum().sum())
print("AI4I missing values  :", ai4i_df.isna().sum().sum())
print("SKAB missing values  :", skab_df.isna().sum().sum())
```
